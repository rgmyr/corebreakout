```python
from tf_unet import unet
from tf_unet import image_util

import numpy as np

from corebreakout.utils import viz_utils

from skimage import io, transform

import random
import glob
```

```python
# pad input images so that output mask is size of input
def pad_batch_to_shape(x, shape):
    btch_pad = (0,0)
    h_pads = tuple([(shape[0] - x.shape[1])//2] * 2)
    w_pads = tuple([(shape[1] - x.shape[2])//2] * 2)
    chan_pad = (0,0)
    pads = (btch_pad, h_pads, w_pads, chan_pad)
    return np.pad(x, pads, 'constant', constant_values=0.0)

# for later convenience (net expects a batch axis)
def to_batch(img):
    return img[np.newaxis,:]

# get a random image from list of image files
def random_img_sample(filelist):
    return to_batch(io.imread(random.choice(filelist)))
```

```python
dbox_path = '/home/administrator/Dropbox/core_data'

model_path = dbox_path+'/saved_models/unet/'             # for saved model ckpt
train_path = dbox_path+'/segmentation/royal_col_augmented/'  # 1000 images
royal_path = dbox_path+'/well_core_data/wos_data/schiehallion-complex/204-19-6/converted/'
```

```python
# create training data generator
img_gen = image_util.ImageDataProvider(train_path+'*.jpeg',
                                      data_suffix='.jpeg', 
                                      mask_suffix='_mask.jpeg',
                                      n_class=2)
```

```python
x_peek, y_peek = img_gen(4) # returns random (img, mask) pair
print(x_peek.shape, y_peek.shape)
```

# Simple training run

```python
# using weighted cross-entropy loss
net = unet.Unet(channels=3, n_class=2, layers=5, features_root=16, cost_kwargs={'class_weights':[1.0, 1.0]})
```

```python
# training run
batch_size = 5  # 8 has memory problems

print(model_path+'test_royal_cols')

trainer = unet.Trainer(net, batch_size=batch_size, optimizer='adam')

path = trainer.train(img_gen,
                     model_path+'test_royal_cols',
                     training_iters=1000//(batch_size),
                     epochs=5, 
                     dropout=0.75)

print(path)
```

```python
# check output on training data
from corebreakout.segmentation.utils import mirror_pad

x, y = img_gen(1)
_x = mirror_pad(x[0,...], (x.shape[1]+194, x.shape[2]+194))
#y = pad_batch_to_shape(y, (512+40, 512+40))

# returns class-wise logits in last axis
pred = net.predict(path, to_batch(_x))

diff = pred[0,...,1]-pred[0,...,0]

thresh = -0.25    # with bad training data, high threshold gives better shape
print(diff.shape)
viz_utils.show_images([x[0,...], y[0,...,1], diff])

#import matplotlib.pyplot as plt
#plt.savefig('test_output.jpg')

print("Input shape: ", x.shape)
print("Output shape: ", pred.shape)
```

# Second try with class-weighted loss

Here we assign higher penalty to the predicting true background as foreground.

Looks like data augmentation would be useful. Predictions look to be partly based on x-coordinate (predicted mask extending above and below what would otherwise be reasonable).

Should also be more careful with mask resizing -- probably better to use some morphological erosion first before (higher-order) resampling.

# using pretty high weighting difference

# looks like predictions should be padded, not interpolated
def get_pads(p, x):
    h_pads = tuple([(x.shape[1] - p.shape[0])//2] * 2)
    w_pads = tuple([(x.shape[2] - p.shape[1])//2] * 2)
    return (h_pads, w_pads)

def background_pad(p, x):
    return np.pad(p, get_pads(p, x), 'constant', constant_values=-1.0)

# check output on training data
x, y = img_gen(1)

# try some flipping

pred = net2.predict(path, x)
