```python
import glob
import random
import utils
import numpy as np
from skimage import io, draw
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams['figure.figsize'] = (12,12)
```

```python
dbox_path = '/home/administrator/Dropbox/core_data'

well_paths = glob.glob(dbox_path+'/well_core_data/wos_data/schiehallion-complex/*')

list(map(lambda s: s.split('/')[-2:], well_paths))
```

```python
def random_well_jpg():
    '''Grab a random UK core photo'''
    well = random.choice(well_paths)
    file = random.choice(glob.glob(well+'/converted/*jpeg'))
    print("Grabbed file %s" % file)
    return io.imread(file)
```
